### 01. PDF Exploration

#### Goal

Load one annual report, extract text from a few pages, and inspect whether
the extraction quality is good enough for later retrieval.

For now, we're not doing chunking, embeddings, RAG or agents.

```python
from pathlib import Path
from pypdf import PdfReader
```

```python
pdf_path = Path("/Users/janhavisingh/personal_projects/financial-analyst-agent/data/raw/tesco_ar25_interactive.pdf")
pdf_path.exists()
```

```python
reader = PdfReader(pdf_path)
print(f"Number of pages: {len(reader.pages)}")
```

```python
pages_to_inspect = 5

extracted_pages = []

for page_number in range(min(pages_to_inspect, len(reader.pages))):
    text = reader.pages[page_number].extract_text()

    extracted_pages.append({
        "page_number": page_number + 1,
        "text": text
    })
```

```python
for page in extracted_pages:
    print("=" * 80)
    print(f"PAGE {page['page_number']}")
    print("=" * 80)
    print(page["text"][:3000])
    print()
```
