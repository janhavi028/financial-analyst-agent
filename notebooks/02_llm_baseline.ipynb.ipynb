```python
import sys

sys.path.append("../src")

from financial_analyst.pdf_processing import extract_pdf_text
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM
from transformers import pipeline
```

```python
report_text = extract_pdf_text("/Users/janhavisingh/personal_projects/financial-analyst-agent/data/raw/tesco_ar25_interactive.pdf", 4, 7)
```

### Get Tokens

```python
model_name = "HuggingFaceTB/SmolLM3-3B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
tokens = tokenizer(
    report_text,
    add_special_tokens=False
)

len(tokens["input_ids"])
```

### Load the model

```python
model_name = "HuggingFaceTB/SmolLM3-3B"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto"
)
model = model.to("mps")
```

```python
question = "What was the company's revenue in the latest financial year?"
```

```python
messages = [
    {
        "role": "system",
        "content": (
            "You are a financial analyst. "
            "Answer the question using only the annual report context provided. "
            "If the answer is not in the context, say that there is not enough information."
        ),
    },
    {
        "role": "user",
        "content": f"""
Annual report context:

{report_text}

Question:
{question}
""",
    },
]
```

```python
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
)

response = generator(
    messages,
    max_new_tokens=100
)

print(response[0]["generated_text"][-1]["content"])
```
